## 2. Prepare Rows for Modeling

Hi again,

We are trying to predict which customers will purchase a book this month.

Can you reformat the data you compiled earlier this week so that it’s ready to be input into a model, with each row representing a customer instead of a purchase?

Thanks!\
Brooke

```python
# What should one row of data look like for the model?
#The row of data for the model should be which customers will purchase in June, with summaries of purchases in April, May
```

```python
#Create a table June Purchases that groups purchases by Customer ID
df_june_purchases = (
    df_q2_sales
        .loc[df_q2_sales['Purchase Date']
        .dt.month == 6]
        .groupby('Customer ID')['Book']
        .count().
        rename('June Purchases')
        .astype(int)
        .reset_index()
)

df_june_purchases.head(5)
```

```python
#Check the data types
df_q2_sales.dtypes
```

```python
# Create a column called 'June Purchases' that sums up all purchases in June, can use mapping(series) or merge with a dataframe
# df_june_purchases = (
#     df_q2_sales.loc[df_q2_sales['Purchase Date'].dt.month == 6].groupby('Customer ID')['Price'].sum().rename('June Purchases')
# )

#Map it to the new column Total Spend, reduces the complexity, 
#reset_index renames the price column to total spend and ensures Book is not an index, this allows it to remain a dataframe
#If you use the map option, omit reset_index, so that it stays a series
#df_q2_sales['June Purchases'] = df_q2_sales['Customer ID'].map(df_june_purchases).fillna(0)

#df_q2_sales.head(5)
```

```python
#Create a column called 'Total Spend' that sums up the prices of all the books that were purchased in April and May.
#Could also do it with a join, Customer ID is the Index to join on
df_total_spend = (
    df_q2_sales.loc[df_q2_sales['Purchase Date'].dt.month.isin([4, 5])].groupby('Customer ID')['Price'].sum().rename('Total Spend').reset_index()
)

df_total_spend.head(50)

#Join option
#df_q2_sales_with_cust = df_q2_sales_with_cust.merge(df_total_spend_map, how='left', on='Customer ID').fillna(0)
```

```python
# Combine the 'June Purchases' and 'Total Spend' columns into a single DataFrame for modelling
df_model = df_total_spend.merge(df_june_purchases, how='left', on='Customer ID').fillna(0)
```

## 3. Prepare Columns for Modeling

Hi again,

Thanks for your help earlier this week! We just learned that we also have to make all of the data numeric before inputting it into a predictive model.

Can you turn the Audience text field into a numeric field?

Thanks!\

```python
#Change the Audience field into a dummy variable, we focus on historic data, so april, may
#Vertical appending
df_april_may_sales = pd.concat([df_april_sales, df_may_sales], axis=0)
#If there are missing values, the NAN values will force the column as float which will cause issues later in joining
df_april_may_sales.head(5)

#After the merge, the NAN values are converting the column type to float
```

```python
#Create subcategories with boolean values to process into the algorithms as they only take in numbers
df_dummy_audience = pd.get_dummies(df_april_may_sales['Audience']).astype('int')
df_dummy_audience.head(5)
```

```python
# Using the Audience dummy variables, create three new columns that contain
# the number of Adult / Children / Teen books purchased by each customer
#Horizontal merge, matches on indexes, because they both come from the same place
df_april_may_grouped = pd.concat([df_april_may_sales['Customer ID'], df_dummy_audience], axis=1)
df_april_may_grouped = df_april_may_grouped.groupby('Customer ID').sum()

df_april_may_grouped.head(5)
```

```python
# Combine the three new columns back with the customer-level data
df_model['June Purchases'] = df_model['June Purchases'].astype('int')
df_model = df_model.merge(df_april_may_grouped, how='left', on='Customer ID')
```

```python
#Model shows total spend, along with sums of subcategories for april and may per customer with total purchases in June
df_model.head(5)
```

```python
# what features could do a good job predicting june purchases?
#Genre
# df_dummy_genre = pd.get_dummies(df_april_may_sales['Genre']).astype('int')
# df_genre_grouped = pd.concat([df_april_may_sales['Customer ID'], df_dummy_genre], axis=1)#Horizontal scaling
# df_genre_grouped = df_genre_grouped.groupby('Customer ID').sum()

#Another way to do the above, hot encoding is already mapping the binary values with get dummies, skips concat as it is in a list
#It does not do hot encoding for Customer ID, since it is already a numeric value, to do it anyways, just wrap it into another pd.get_dummies
df_genre_grouped = (
    pd.get_dummies(df_april_may_sales[['Customer ID', 'Genre']], dtype='int')
    .rename(columns=lambda col: col.removeprefix('Genre_'))
    .groupby('Customer ID').sum()
    .reset_index()
)

df_genre_grouped.head(5)

#df_genre_grouped.loc[df_genre_grouped['Customer ID'] == 102].head(5)
```

```python
df_genre_grouped = df_genre_grouped.groupby('Customer ID').sum().reset_index()
df_genre_grouped.head(5)
```

```python
#Purchase Location
# df_dummy_purchase_loc = pd.get_dummies(df_april_may_sales['Purchase Location']).astype('int')

# df_purchase_loc_grouped = pd.concat([df_april_may_sales['Customer ID'], df_dummy_purchase_loc], axis=1) #Horizantal scaling
# df_purchase_loc_grouped = df_purchase_loc_grouped.groupby('Customer ID').sum().reset_index()

#Another way to do the above, hot encoding is already mapping the binary values with get dummies, skips concat as it is in a list
#It does not do hot encoding for Customer ID, since it is already a numeric value, to do it anyways, just wrap it into another pd.get_dummies
df_purchase_loc_grouped = (
    pd.get_dummies(df_april_may_sales[['Customer ID', 'Purchase Location']], dtype='int')
    .groupby('Customer ID').sum()
    .rename(columns=lambda col: col.removeprefix('Purchase Location_'))
    .reset_index()
)

df_purchase_loc_grouped.head(5)
```

```python
# add it to the dataframe for modeling
df_model = df_model.merge(df_genre_grouped, how='left', on='Customer ID').merge(df_purchase_loc_grouped, how='left', on='Customer ID')
```

```python
#Add percentages of in person
#Series
s_in_person = (df_model['In Person'] / (df_model['In Person'] + df_model['Online']) * 100).rename('In Person %')
s_in_person.head(5)
```

```python
#Add percentages of online purchases
#Series
s_online = (df_model['Online'] / (df_model['Online'] + df_model['In Person']) * 100).rename('Online %')
s_online.head(5)
```

```python
#Add them in the model
df_model = pd.concat([df_model, s_in_person, s_online], axis=1)
```

```python
#Round the values to 3 float numbers
df_model['In Person %'] = round(df_model['In Person %'], 2)
df_model['Online %'] = round(df_model['Online %'], 2)
```

```python
#Remove the Customer ID, as it is no longer relevant for the model, can track customers based on the index if needed
df_model = df_model.drop('Customer ID', axis=1)
```
