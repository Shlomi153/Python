## Read in the data

```python
# import the data 
import pandas as pd

data_folder = r''happiness_survey_file = 'happiness_survey_data.csv'
```

```python
df_happiness_survey = pd.read_csv(data_folder + r'\\' + happiness_survey_file)
```

```python
# there are 2089 rows and 6 columns of data
df_happiness_survey.shape
```

```python
# the years range from 2005 - 2021
# happiness scores range from 2 - 8
df_happiness_survey.describe()
```

```python
df_happiness_survey.head(3)
```

```python
#Round the numbers to 2 decimal points to make it easier to read
df_happiness_survey = df_happiness_survey.round({'happiness_score':2, 'social_support':2, 'freedom_to_make_life_choices': 2, 'healthy_life_expectancy_at_birth': 2})
```

```python
df_happiness_survey.head(2)
```

## Explore the data

```python
# Filter out any data before 2010 and after 2019
df_happiness_survey = df_happiness_survey.loc[(df_happiness_survey['year'] >= 2010) & (df_happiness_survey['year'] <= 2019)]
```

```python
# Group the data by country and calculate the maximum happiness score for each one
df_happiness_survey.head(3)
```

```python
# Sort the grouped countries by happiness score and return the top five
happiness_scores_max = df_happiness_survey.groupby('country_name')['happiness_score'].max()
```

```python
happiness_scores_max.head(5)
```

```python
# Group the data by country and calculate the average happiness score for each one
happiness_scores_avg = df_happiness_survey.groupby('country_name')['happiness_score'].mean().round(2)
```

```python
happiness_scores_avg.head(5)
```

```python
# Sort the grouped countries by happiness score and return the top five
sort_happiness_scores_max = happiness_scores_max.sort_values(ascending=False)
```

```python
sort_happiness_scores_avg = happiness_scores_avg.sort_values(ascending=False)
```
